This notebook demonstrates:
- Multi-dimensional tensor creation and GPU/CPU device management

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

print(f'PyTorch Version: {torch.__version__}')
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device: {device}')
```

### 1. Tensor Operations & Broadcasting
Simulate field soil nutrient sensor readings (Nitrogen, Phosphorus, Potassium):


```python
sensors = torch.tensor([
    [120.5, 45.0, 60.2],
    [90.0, 30.5, 40.0],
    [150.2, 55.0, 75.8],
    [80.0, 25.0, 35.0]
], dtype=torch.float32, device=device)

offset = torch.tensor([-5.0, 2.5, -1.0], device=device)
calibrated = sensors + offset
print('Calibrated Readings (Broadcasted):\n', calibrated)
```

### 2. Autograd & Reverse-Mode Differentiation
Formulating dynamic transpiration rate gradient computation:


```python
w1 = torch.tensor(2.0, requires_grad=True)
w2 = torch.tensor(0.15, requires_grad=True)
b = torch.tensor(0.5, requires_grad=True)

solar = torch.tensor([0.8, 1.2, 0.5, 1.5])
temp = torch.tensor([24.0, 28.5, 22.0, 31.0])
y_true = torch.tensor([4.5, 7.2, 3.1, 9.8])

y_pred = w1 * (solar ** 2) + w2 * temp + b
loss = torch.mean((y_pred - y_true) ** 2)
loss.backward()

print(f'Loss: {loss.item():.4f}')
print(f'dL/dw1: {w1.grad.item():.4f} | dL/dw2: {w2.grad.item():.4f} | dL/db: {b.grad.item():.4f}')
```
